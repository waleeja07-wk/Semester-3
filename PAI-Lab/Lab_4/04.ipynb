## `for` loop: count is known

"A good programmer is a lazy programmer!"

```python
for ingredient in 'sugar', 'milk', 'tea leaves', 'water': 
    print("Put", ingredient, 'in.')
```

## `range` + `break`

`break` → leave the loop now.

```python
n = 14
for i in range(1, n):   # loop over a few numbers 
        if i % 5 == 0:
            break
            
        print(i)

print("done")
```

```python
# continue = skip this round, go next
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# range(start, stop, step)
for i in range(2, 11, 2):
    print(i)
```

```python
# sum 1..n with a loop
total = 0
for i in range(1, 6):
    total = total + i
print(total)  # 15
```

## `while` loop: stop when condition fails

Use when you know *when* to stop, not *how many* times.

```python
n = 0.13 
counter = 0

while n < 1 :
    counter = counter + 1
    print(n) 
    n = n + 0.13 

print("counter: ", counter)
print('Done.')
```

```python
# we can also count how many there were by saving the count
n = 0.13 
i = 0         # our counter 

while n < 1: 
    i = i + 1
    print(n) 
    n = n + 0.13 

print(i) 
print('Done.')
```

```python
# we can also count how many there were by saving the count
n = 0.13 
i = 1         # our counter 

while i < 11: 
    i = i + 1
    print(n) 
    n = n + 0.13 

print(i) 
print('Done.')
```

```python
# countdown
x = 5
while x > 0:
    print(x)
    x = x - 1
print("Go!")
```

## Factorial

`4! = 4 × 3 × 2 × 1 = 24`

```python
fac = 1    # typically, the iteration variable is called i or j or k 

i = 1 

while i <= 4 : 
    fac = fac * i
    i = i + 1 
    
print(fac) 
```

```python
# same idea with for
fac = 1
for i in range(1, 5):
    fac = fac * i
print(fac)
```
