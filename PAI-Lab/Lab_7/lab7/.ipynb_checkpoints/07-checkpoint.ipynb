## `from module import *`: avoid it

`*` copies **every** name from the module into our program. If we already have a name with the same spelling, it is silently replaced.

````python
def mean(values):
    return "my own mean function"

from statistics import *

mean([1, 2, 3])   # our mean() was replaced by statistics' mean() without any warning
````

Prefer `import math` or import specific names.

## Python `import` vs C/C++ `#include`

In C++ you wrote `#include <cmath>`. Both give you ready-made code, but they work differently.

| | C/C++ `#include` | Python `import` |
|---|---|---|
| Done by | Preprocessor, before compiling | Interpreter, while running |
| What happens | File contents are **copy-pasted** in | Module is **loaded and run once** |
| Names | Come straight into your file | Stay inside the module's **namespace** |
| Use | `sqrt(25)` | `math.sqrt(25)` |

```cpp
#include <iostream>
#include <cmath>
using namespace std;

int main() {
    cout << sqrt(25);
}
```

Because each module keeps its own names, a module's function and a built-in function can have the same name without a clash:

````python
print(pow(2, 3))        # built-in pow
print(math.pow(2, 3))   # math's pow always gives a float
````

We will see "loaded only once" for ourselves when we write our own module.

- The **standard library** is a large set of modules that comes with Python. No installation is needed, but you must `import` them first.

| Module | Used for |
|---|---|
| `math` | mathematical functions |

### `random`

A computer cannot "flip a coin" by itself; it can only follow instructions. So `random` uses a **formula** that produces a long chain of numbers that *look* random. These are called **pseudo-random** numbers.

Think of it like this:
- Python has a **very long, fixed list** of mixed-up numbers.
- Every time you ask for a random value, it takes the **next** number from that list.
- The **seed** decides **where in the list you start**. (More on the seed below.)

| Function | Gives you | Example result |
|---|---|---|
| `random.random()` | a float from 0 up to (not including) 1 | `0.5714` |
| `random.randint(a, b)` | an int from `a` to `b`, **both included** | `randint(1, 6)` → `4` |
| `random.uniform(a, b)` | a float between `a` and `b` | `uniform(1, 5)` → `3.27` |
| `random.choice(seq)` | **one** item from a sequence | `'Sadia'` |
| `random.sample(seq, k)` | a **new list** of `k` **different** items | `['Zara', 'Ali']` |
| `random.choices(seq, k=n)` | a new list of `n` items, **repeats allowed** | `['Ali', 'Ali', 'Omar']` |
| `random.shuffle(lst)` | nothing; it **mixes the list itself** | `None` |
| `random.seed(n)` | nothing; it **fixes the starting point** | `None` |

````python
import random
````

#### `random.random()`: a float between 0 and 1

The result can be `0.0` but is **never** `1.0`. Run the cell a few times: the value changes every time.

````python
random.random()
````

````python
for i in range(5):
    print(random.random())
````

It looks useless, but it is the building block for the others. Two common uses:

````python
# 1) scale it: a float between 0 and 100
print(random.random() * 100)

# 2) "this happens with a 30% chance"
# random.random() is below 0.3 roughly 30 times out of 100
if random.random() < 0.3:
    print("You won a prize!")
else:
    print("Better luck next time")
````

#### `random.randint(a, b)`: a whole number from `a` to `b`

Careful: unlike `range(1, 6)`, `randint(1, 6)` **includes 6**. Perfect for a dice.

````python
random.randint(1, 6)
````

````python
rolls = []
for i in range(10):
    rolls.append(random.randint(1, 6))
print(rolls)          # 1 and 6 can both appear; values can repeat
````

````python
random.uniform(1.5, 2.5)   # like randint, but gives a float (e.g. a random height in metres)
````

#### `random.choice(seq)`: pick ONE item

Works on any sequence: list, tuple, string, range.

````python
names = ["Ali", "Sadia", "Hamza", "Zara", "Omar"]

print(random.choice(names))
print(random.choice("PYTHON"))     # a random letter
print(random.choice(range(10, 20)))
````

Every call is independent, so the same item can come up again:

````python
for i in range(5):
    print(random.choice(names))
````

#### `random.sample(seq, k)`: pick `k` DIFFERENT items

Think of a **lucky draw**: once a name slip is pulled out of the box, it does not go back in. So nobody can be picked twice.
- It returns a **new list**.
- The original list is **not changed**.

````python
winners = random.sample(names, 3)

print(winners)    # 3 different names
print(names)      # original list is unchanged
````

````python
random.sample(names, 6)   # error: cannot pick 6 different items from only 5
````

Compare with `random.choices` (with an **s**): the slip **goes back** in the box, so repeats are allowed. Like rolling a dice many times.

````python
print(random.sample(names, 5))       # never repeats: every name exactly once
print(random.choices(names, k=5))    # repeats possible
````

#### `random.shuffle(lst)`: mix up a list, like shuffling cards

- It changes the list **itself** (in place).
- It **returns nothing** (`None`), just like `list.sort()`.

````python
cards = ["A", "2", "3", "4", "5", "6"]

random.shuffle(cards)
print(cards)      # same items, new order
````

````python
# a very common mistake
cards = ["A", "2", "3", "4", "5", "6"]

mixed = random.shuffle(cards)
print(mixed)      # None! shuffle() returned nothing
print(cards)      # the shuffling happened here
````

````python
order = ("Ali", "Sadia", "Zara")
random.shuffle(order)    # error: a tuple cannot be changed, so it cannot be shuffled
````

Need a shuffled **copy** and want to keep the original? Use `sample` with `k` equal to the length:

````python
original = ["Ali", "Sadia", "Hamza", "Zara"]
shuffled_copy = random.sample(original, len(original))

print(original)
print(shuffled_copy)
````

#### `random.seed(n)`: making "random" repeatable

Remember the long, fixed list of numbers? The **seed is the starting point** in that list.

- **No seed** → Python picks a starting point by itself (from the clock / the system), so you get **different** results every run.
- **Same seed** → **same** starting point → exactly the **same** results, on every run and on every computer.
- **Different seed** → a different starting point → different results.

Analogy: the seed is like the **page number of a book of random numbers**. If we both open page 10, we read the same numbers.

Run the next cell twice: the numbers change.

````python
print(random.randint(1, 100), random.randint(1, 100), random.randint(1, 100))
````

Now with a seed. Run it as many times as you want: **always the same three numbers**.

````python
random.seed(10)
print(random.randint(1, 100), random.randint(1, 100), random.randint(1, 100))
````

````python
random.seed(10)      # back to the same page
print(random.randint(1, 100), random.randint(1, 100), random.randint(1, 100))
````

````python
random.seed(99)      # a different page -> different numbers
print(random.randint(1, 100), random.randint(1, 100), random.randint(1, 100))
````

Important points about the seed:

1. The seed does **not** give the same number every time. It gives the same **sequence**: the 1st, 2nd and 3rd numbers differ from each other, but they are the same on every run.
2. Call `seed` **once**, at the top of the program. Calling it again **restarts** the sequence from the beginning.
3. It affects **all** `random` functions: `random()`, `randint`, `choice`, `sample`, `shuffle`, ...
4. The number itself does not matter. `42` is just a popular choice.

````python
# point 1: one seed, then several calls -> the numbers differ, but the whole list repeats on every run
random.seed(5)
for i in range(5):
    print(random.randint(1, 100), end=" ")
````

````python
# point 2: calling seed before EVERY call restarts the sequence each time -> the same number again and again
for i in range(5):
    random.seed(5)
    print(random.randint(1, 100), end=" ")
````

````python
# point 3: the seed controls choice, sample and shuffle too
random.seed(7)
names = ["Ali", "Sadia", "Hamza", "Zara", "Omar"]
print(random.choice(names))
print(random.sample(names, 2))
random.shuffle(names)
print(names)
````

Why would we want "random" numbers that repeat?
- **Testing / debugging:** if a bug appears with random data, a seed lets you reproduce it.
- **Checking work:** in the lab tasks everyone uses the same seed, so everyone gets the same output.
- **Machine learning (later):** the data is split and shuffled randomly. A seed makes your results reproducible.

When should you **not** use a seed? When you want real unpredictability, e.g. a game. (For passwords and security, Python has a separate `secrets` module. `random` is predictable, so never use it for those.)

#### Putting it together: a lucky draw

````python
random.seed(2026)

students = ["Ali", "Sadia", "Hamza", "Zara", "Omar", "Ahmad", "Waqas"]

prize_winners = random.sample(students, 3)    # 3 different winners
print("Winners:", prize_winners)

random.shuffle(prize_winners)                 # random order for the stage
print("Stage order:", prize_winners)

lucky_number = random.randint(1, 100)
print("Lucky number:", lucky_number)

bonus = random.random() < 0.5                 # 50% chance
print("Bonus round:", bonus)
````

### `collections.Counter`

In Lab 06 we counted votes with a dictionary and a loop. `Counter` does it in one line.

````python
from collections import Counter

votes = ["Ali", "Sadia", "Ali", "Hamza", "Sadia", "Ali", "Zara"]
counts = Counter(votes)

print(counts)
print(counts["Ali"])
print(counts.most_common(1))   # [(name, count)] of the winner
````

## Creating your own module

Any `.py` file is a module. Let's create `geometry.py` next to this notebook.

`%%writefile filename` at the top of a cell **saves** the cell into a file instead of running it.

````python
%%writefile geometry.py
"""Small helper functions for shapes."""
import math

UNITS = "cm"

print("geometry module is loading...")


def circle_area(r):
    """Return the area of a circle of radius r."""
    return math.pi * r * r


def rectangle_area(length, width):
    """Return the area of a rectangle."""
    return length * width


def hypotenuse(a, b):
    """Return the longest side of a right-angled triangle."""
    return math.sqrt(a * a + b * b)
````

````python
import geometry
````

````python
print(geometry.circle_area(3))
print(geometry.UNITS)
````

````python
import geometry     # nothing is printed this time!
````

A module runs **only the first time** it is imported. Python remembers it, so the next `import` reuses it. This is the "loaded once" difference from `#include`.

````python
from geometry import rectangle_area

rectangle_area(4, 5)
````

````python
import geometry as geo

geo.hypotenuse(3, 4)
````

````python
help(geometry)     # our docstrings become the module's help
````

If you edit the module's file, a running notebook will not see the change. Restart the kernel, or reload it:

````python
import importlib

importlib.reload(geometry)
````

## `if __name__ == "__main__":`

Every module has a variable `__name__`.
- File **run directly** (`python converter.py`) → `__name__` is `"__main__"`
- File **imported** → `__name__` is the module's name, `"converter"`

So code under this `if` runs only when the file is run directly. It is the right place for tests.

````python
%%writefile converter.py
"""Unit conversions."""


def c_to_f(c):
    """Celsius to Fahrenheit."""
    return c * 9 / 5 + 32


def km_to_miles(km):
    """Kilometres to miles."""
    return km * 0.621371


if __name__ == "__main__":
    print("Testing converter...")
    print(c_to_f(100))       # 212.0
    print(km_to_miles(10))   # 6.21371
````

````python
!python converter.py
````

`!` runs a terminal command from the notebook. Here the file ran directly, so the tests ran.

````python
import converter      # imported: the tests do NOT run

converter.c_to_f(37)
````

````python
print(converter.__name__)
print(__name__)        # this notebook is the program being run
````

### Common mistake: naming your file like a standard module

If you save your own file as `random.py`, then `import random` loads **your** file (the current folder is searched first), and `random.randint(...)` stops working. Never name your files `random.py`, `math.py`, `statistics.py` ...

## Packages

When a project grows, related modules are grouped into a folder. That folder is a **package**.

```text
university/
├── __init__.py      runs once, the first time the package is imported
├── grades.py
└── fees.py
```

````python
os.makedirs("university", exist_ok=True)    # create the folder (no error if it already exists)
````

````python
%%writefile university/__init__.py
"""university package."""

NAME = "City University"
print("Loading the university package...")
````

````python
%%writefile university/grades.py
"""Grading rules."""


def grade(marks):
    """Return the letter grade for marks out of 100."""
    if marks >= 85:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 50:
        return "C"
    return "F"
````

````python
%%writefile university/fees.py
"""Fee rules."""

PER_CREDIT = 8000


def semester_fee(credits):
    """Return the fee for the given credit hours."""
    return credits * PER_CREDIT


def apply_scholarship(fee, percent):
    """Return the fee after taking percent off."""
    return fee - fee * percent / 100
````

````python
import university

university.NAME
````

````python
from university import grades

for m in [91, 74, 45]:
    print(m, grades.grade(m))
````

````python
from university.fees import semester_fee, apply_scholarship

fee = semester_fee(18)
print(fee, apply_scholarship(fee, 25))
````

### Sub-packages

A package can contain another package (a folder inside the folder). The dots in the import follow the folders.

````python
os.makedirs("university/hostel", exist_ok=True)
````

````python
%%writefile university/hostel/__init__.py
# this file can stay empty: it just marks the folder as a package
````

````python
%%writefile university/hostel/rooms.py
def room_rent(sharing):
    """Monthly rent: fewer students in a room means more rent."""
    rents = {1: 25000, 2: 15000, 3: 10000}
    return rents.get(sharing, 8000)
````

````python
from university.hostel.rooms import room_rent

room_rent(2)
````

## `pip`: installing packages written by others

The standard library is big, but not everything is in it. Thousands of packages are published on **PyPI** (Python Package Index). `pip` downloads and installs them.

pip commands are typed in the **terminal**, e.g. `pip install tabulate`.

In a notebook, write `%pip` instead. It installs into the same Python that the notebook is running. (`!pip` runs the terminal's pip, which may belong to a **different** Python, so the package may install but still fail to import.)

````python
%pip --version
````

````python
%pip list
````

````python
%pip install tabulate
````

````python
from tabulate import tabulate

rows = [("Sadia", 85, "A"), ("Ali", 74, "B"), ("Hamza", 45, "F")]
print(tabulate(rows, headers=["Name", "Marks", "Grade"]))
````

````python
%pip show tabulate
````

Other useful commands:

| Command | What it does |
|---|---|
| `pip install numpy==2.5.3` | install a specific version |
| `pip install --upgrade numpy` | upgrade to the newest version |
| `pip uninstall -y tabulate` | remove a package (`-y` answers "yes", as a notebook cannot type it) |
| `pip freeze > requirements.txt` | save the installed packages and versions in a file |
| `pip install -r requirements.txt` | install everything listed in that file (e.g. on another computer) |
