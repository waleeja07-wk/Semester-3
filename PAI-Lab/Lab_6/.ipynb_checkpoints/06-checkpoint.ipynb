# Lab 06: Lists, Tuples, Sets and Dictionaries

A **sequence** is an ordered collection of values. Lists, tuples, strings and ranges are all sequences.

All sequences share two behaviours:
- **length** → `len(s)`
- **element selection** → `s[index]`

Let's do some interesting stuff with lists. A `for` loop visits each item one by one.

```python
def make_tea(ing):
    for i in ing:
        print("Put", i, "in.")


ingredients = ['water', 'milk', 'cinnamon', 'tea leaves']
make_tea(ingredients)
```

Counting how many times a value appears in a list:

```python
seq = [1, 2, 8, 4, 6, 8, 9, 15]
```

```python
def count(s, value):
    """Count how many times value occurs in s"""
    counter = 0

    for i in s:
        if i == value:
            counter = counter + 1

    return counter
```

```python
count(seq, 88)
```

```python
count(seq, 8)
```

```python
seq.count(8)   # lists already have a built-in method for this
```

## Nested sequences

Just as we had nested expressions, we can have nested sequences. Here, each item of `pairs` is itself a list of two values.

```python
pairs = [
    [1, 20],
    [2, 30],
    [3, 40],
    [4, 50]
]
```

```python
pairs[1][0]     # second pair, first value
```

Since every item is a pair, the loop can **unpack** it into `x` and `y` directly.

```python
for x, y in pairs:
    print("x =", x, "and y =", y)
```

Count the pairs whose two values are the same, and remember their positions:

```python
pairs = [[1, 1], [2, 2], [2, 3], [4, 4], [8, 8]]
```

```python
same_count = 0
index = 0
same_indices = []

for x, y in pairs:
    if x == y:
        same_count += 1     # same_count = same_count + 1
        same_indices.append(index)
    index += 1

print(same_count)
print(same_indices)
```

## List comprehensions

A short way to build a new list from an existing one. Both cells below give the same result.

```python
nums = [1, 2, 3, 4, 5, 6, 101, 105, 1000]
evens = []

for i in nums:
    if i % 2 == 0:
        evens.append(i)     # append in output

print(evens)
```

```python
evens = [i for i in nums if i % 2 == 0]
print(evens)
```

```python
n = 11
[x for x in range(1, n + 1) if n % x == 0]     # all divisors of n
```

## Tuples - Immutable sequences

A tuple is almost the same as a list, but it **cannot be changed** after it is created. Use it for fixed data.

```python
digits = (1, 8, 2, 8)
```

```python
len(digits)
```

```python
digits[1]
```

```python
digits[1:3]
```

```python
digits[0] = 11   # except for this!
```

A tuple keeps related values together as one record, and it can be **unpacked** into variables:

```python
student = ("Ali", 21, "BSCS")      # a fixed record

name, age, program = student       # unpacking
print(name, age, program)
```

A list of tuples is a common way to store many records. We loop over it just like `pairs`:

```python
results = [("Ali", 14), ("Sara", 18), ("Hamza", 9)]

for name, marks in results:
    print(name, "got", marks)
```

## Sets

A set is an **unordered** collection of **unique** values, written with `{}`. Duplicates are removed automatically, so a set is useful when we only care about *which* values exist.

```python
data = {1, 2, 2, 3, 3, 4}
print(data)
```

Converting a list to a set is the quickest way to remove duplicates:

```python
digits = [1, 8, 2, 8]

unique = set(digits)
print(unique)
print(len(unique))
```

```python
unique[0]     # error: a set has no order, so no indexing
```

Adding and removing elements:

```python
fruits = {"apple", "banana"}

fruits.add("orange")
fruits.add("apple")        # already present, so nothing changes
fruits.remove("banana")

print(fruits)
print("apple" in fruits)   # membership check
print(sorted(fruits))      # sorted() gives the values as an ordered list
```

Note: `{}` creates an empty **dictionary**, so an empty set is written as `set()`.

```python
seen = set()
seen.add(5)
print(seen)
```

### Set operations

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A | B)   # Union: in A or in B
print(A & B)   # Intersection: in both
print(A - B)   # Difference: in A but not in B
```

## Dictionaries

A dictionary stores **key → value** pairs. We look up a value by its key, not by its position, just like looking up a word in a glossary.

```python
# recall lists: we select items by position
lst = [1, 2, 3, 4]
lst[0]
```

```python
d = {
    "package": "A package is a module that can be added to any program",
    "random": "Any data or information that has no order",
    "unary operator": "An operator that takes only one value for its operation"
}
```

```python
d['random']
```

Assigning to a new key adds it. Assigning to an existing key updates its value.

```python
d['variable'] = "A name that refers to a value"
```

```python
d
```

```python
d['python']     # error: this key does not exist
```

```python
d.get('python', 'Not found')   # get() returns a default value instead of an error
```

```python
del d['variable']   # remove a key and its value
list(d.keys())
```

## Iterating over dictionaries

```python
for k in d:
    v = d[k]
    print(k, "=", v)
```

```python
'package' in d      # checks the keys
```

```python
'else' in d
```

```python
if 'else' not in d:   # you can move the 'not'
    print("The keyword 'else' is not in d... how can this be?")
else:
    print("This keyword was found.")
```

```python
list(d.items())       # returns a list of tuples!
```

Now, we can iterate over them the same way we iterated over `pairs`:

```python
for k, v in d.items():
    print(k, "=", v)
```

## Counting with a dictionary

A very common use: count how many times each value appears. The value becomes the **key**, and its count becomes the **value**.

```python
digits = [1, 8, 2, 8]
counts = {}

for x in digits:
    if x in counts:
        counts[x] = counts[x] + 1   # seen before: increase its count
    else:
        counts[x] = 1               # first time: start at 1

print(counts)
```
