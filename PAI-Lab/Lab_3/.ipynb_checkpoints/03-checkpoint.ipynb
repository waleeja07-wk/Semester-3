# Lab 03: Making Decisions in Python

In everyday life, we make decisions all the time based on conditions:
- If you have your student ID card, you can enter school.
- If it is raining, you take an umbrella.
- If the traffic light is green, you go.

In Python, we use **conditional statements** (`if`, `elif`, `else`, and `match-case`) to make decisions in code.

---
## 1. Simple Decision: The `if` Statement

The `if` statement checks a condition. If the condition is True, Python runs the code inside it. If it is False, Python skips it.

### Syntax
```python
if condition:
    # Do something here
```

Let's check if it is raining outside.

````python
# Example 1: Checking if it is raining
is_raining = True

if is_raining:
    print("It is raining! Remember to take your umbrella.")

print("Have a nice day!")
````

---
## 2. Choosing Between Two Options: The `if-else` Statement

When you have two choices, use `if-else`. If the condition is True, the `if` block runs. Otherwise, the `else` block runs.

### Syntax
```python
if condition:
    # Code if True
else:
    # Code if False
```

Let's check if a student passed an exam with a passing mark of 50.

````python
# Example 2: Checking exam pass or fail
marks = 65

if marks >= 50:
    print("Marks:", marks)
    print("Result: You passed the exam!")
else:
    print("Marks:", marks)
    print("Result: You failed. Keep trying!")
````

---
## 3. Multiple Choices: The `if-elif-else` Statement

When you have more than two choices, use `elif` (which means "else if"). Python checks each condition one by one from top to bottom.

Let's check a student score and assign a grade (A, B, C, D, or F).

````python
# Example 3: Grading system
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
````

---
## 4. Combining Conditions (`and`, `or`, `not`)

Sometimes you need to check more than one thing at the same time:
- `and`: Both things must be True.
- `or`: At least one thing must be True.
- `not`: Reverses True to False, or False to True.

Let's check school gate permission using mask, ID card, and uniform checks.

````python
# Example 4: School gate security check
has_mask = True
has_id_card = True
has_uniform = True

if has_mask and has_id_card and has_uniform:
    print("All requirements met. You are allowed to enter!")
else:
    print("Entry denied. Please check your mask, ID card, or uniform.")
````

---
## 5. Nested `if` Statements (An `if` inside an `if`)

You can place an `if` statement inside another `if` statement. This is called a nested `if`.

Let's check school gate entry step-by-step to tell the student exactly what is missing.

````python
# Example 5: Detailed school gate check
has_mask = True
has_id_card = False
has_uniform = True

if has_mask:
    if has_id_card:
        if has_uniform:
            print("You are allowed to enter.")
        else:
            print("Your uniform is missing. You are not allowed.")
    else:
        print("You don't have an ID card. You are not allowed.")
else:
    print("You don't have a mask. You are not allowed.")
````

---
## 6. One-Line Decision (Ternary Operator)

If you have a very short `if-else` decision, you can write it on one line.

### Syntax
`result = value_if_true if condition else value_if_false`

````python
# Example 6: Pass or Fail result
score = 75

result = "Passed" if score >= 50 else "Failed"
print("Exam Result:", result)
````

---
## 7. Matching Patterns: The `match-case` Statement

Python has a feature called `match-case`. It compares a value against different patterns. It is very easy to read when you have many fixed choices.

### Syntax
```python
match variable:
    case value1:
        # Action 1
    case value2:
        # Action 2
    case _:
        # Default action if nothing matches
```

Let's make a simple animal sound maker.

````python
# Example 7: Animal sound maker using match-case
animal = "dog"

match animal:
    case "dog":
        print("The dog says: Woof woof!")
    case "cat":
        print("The cat says: Meow meow!")
    case "duck":
        print("The duck says: Quack quack!")
    case "cow":
        print("The cow says: Moo moo!")
    case _:
        print("Unknown animal sound!")
````
