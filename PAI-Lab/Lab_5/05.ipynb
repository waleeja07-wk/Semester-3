# Functions

## 1. Defining and Calling Functions

We define a function with the `def` command. The indented block is the function body.

```python
def square(x):
    return x * x  # return the squared value
```

A function runs when we call it. The value returned by the function can be stored in a variable.

```python
v = square(5)
v
```

### Parameters and return values

Parameters are the names in the function definition. When we call the function, the supplied values become arguments. A function can also call another function.

```python
def sum_squares(x, y):
    return square(x) + square(y)

result = sum_squares(3, 4)
result
```

## 2. Scope, Defaults, `lambda`, and Docstrings

### Scope

Scope describes where a variable can be used. A parameter and a variable assigned inside a function are local to that function. A name created outside the function is global to the notebook.

```python
def f(x):
    y = 1
    x = x + y  # this changes the local x, not the outer x
    print('. x =', x)
    print('. p =', p)
    return x

p = 10
x = 3
z = f(x)

print('x =', x)
print('z =', z)
```

The `x` inside `f` is a local parameter, so changing it does not change the outer `x`. The local variable `y` exists only while `f` is running.

### Default arguments

A default value lets callers omit an argument when the usual value is meaningful.

```python
def sqrt(num, guess):
    print('Finding square root of', num, 'starting with guess:', guess)

sqrt(36, 1)
```

Without a default, `sqrt(36)` would need to supply the missing `guess` argument. We can make `1.0` the default.

```python
def sqrt(num, guess=1.0):
    print('Finding square root of', num, 'starting with guess:', guess)

sqrt(36)
sqrt(25, 3)
```

A default argument is also handy for simple reusable messages.

```python
def greet(name, greeting='Hello'):
    return f'{greeting}, {name}!'

print(greet('Mina'))
print(greet('Mina', 'Good morning'))
```

### Lambda expressions

A `lambda` creates a small anonymous function. It is useful when the function is short and only needed in one place. The `def` version beside it does the same thing with a named function.

```python
def square_with_def(x):
    return x * x

square_with_lambda = lambda x: x * x

print(square_with_def(5))
print(square_with_lambda(5))
```

### Designing and documenting functions

Functions are abstractions, so their operation should depend as little as possible on the surrounding environment. Use meaningful names and parameter names, and document functions that others will use.

```python
def pressure(v, t, n):
    """Compute the pressure in pascals of an ideal gas.

    Applies the ideal gas law.

    v -- volume of gas, in cubic meters
    t -- absolute temperature in degrees kelvin
    n -- particles of gas
    """
    k = 1.38e-23
    return n * k * t / v
```
