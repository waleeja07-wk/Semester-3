## 1. What are Collections in Python?

A **Collection** is a data structure that can store multiple values in a single variable. Python provides several collection types such as Lists, Tuples, Sets, and Dictionaries.

### Key Points

- Store multiple items together.
- Reduce the need for many separate variables.
- Allow easy manipulation of grouped data.
- Can contain different data types.
- Widely used in AI, Data Science, and General Programming.


### Example

````python
numbers = [10, 20, 30, 40, 50]
print(numbers)
````

# 2. Lists in Python


A **List** is an ordered, mutable (changeable) collection of items enclosed in square brackets `[]`.

### Characteristics of Lists

- Ordered collection.
- Mutable (can be modified).
- Allows duplicate values.
- Can store different data types.
- Supports indexing and slicing.

**Note:** By `Ordered collection` we mean that the elements in a list maintain a specific order, and each element has a position (index). Because **ordered** is a property of the data structure, not something unique to lists.

### Example

````python
student = ["Ali", 20, 3.8, True]
print(student)
````

# 3. Creating Lists

### Empty List

````python
my_list = []
print(my_list)
````

### List of Integers

````python
numbers = [1, 2, 3, 4, 5]
numbers
````

### List of Strings

````python
names = ["Ali", "Sara", "Ahmed"]
names
````

### Mixed Data Types

````python
data = [10, "Python", 3.14, True]
print(data)
````

# 4. Accessing List Elements

Lists use **zero-based indexing**.

````python
names = ["Ali", "Sara", "Ahmed"]

print(names[0])
print(names[1])
print(names[2])
````

### Negative Indexing

````python
names = ["Ali", "Sara", "Ahmed"]

print(names[-1])
print(names[-2])
````

# 5. List Slicing

### Syntax

```python
list[start:end:step]
```

### Example

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
print(numbers[::2])
````

### Modifying List Elements

````python
numbers = [10, 20, 30]

numbers[1] = 99

print(numbers)
````

# 6. Common List Operations

## Length of List

````python
numbers = [10, 20, 30, 40]

print(len(numbers))
````

## Concatenation

````python
list1 = [1, 2]
list2 = [3, 4]

result = list1 + list2
print(result)
````

## Repetition

````python
numbers = [1, 2]

print(numbers * 3)
````

## Membership Operator

````python
numbers = [10, 20, 30]

print(20 in numbers)
print(100 in numbers)
````

# 7. Important List Methods

## append()

Adds an item at the end.

````python
numbers = [10, 20, 30]

numbers.append(40)
print(numbers)
````

## insert()

Adds an item at a specific position.

````python
numbers = [10, 20, 30]

numbers.insert(1, 15)
print(numbers)
````

## remove()

Removes the first matching value.

````python
numbers = [10, 20, 30, 20]

numbers.remove(20)

print(numbers)
````

## pop()

Removes an element by last index.

````python
numbers = [10, 20, 30]
numbers.pop()
print(numbers)
````

## sort()

Sorts the list.

````python
numbers = [40, 10, 30, 20]
numbers.sort()
print(numbers)
````

## reverse()

Reverses the list.

````python
numbers = [10, 20, 30]
numbers.reverse()
print(numbers)
````

## clear()

Removes all elements.

````python
numbers = [10, 20, 30]
numbers.clear()
print(numbers)
````

# 8. Iterating Through Lists

## Using for Loop

````python
numbers = [10, 20, 30, 40]
for n in numbers:
    print(n)
````

````python
numbers = [10, 20, 30, 40]
for n in numbers:
    print(n, numbers[n])
````

## Using Index

````python
numbers = [10, 20, 30]
for i in range(len(numbers)):
    print(i, numbers[i])
````

# 9. Nested Lists

A list can contain other lists.

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[1][2])
````

# 10. Lists vs C/C++ Arrays

| Feature | Python List | C/C++ Array |
|---|---|---|
| Size | Dynamic | Fixed |
| Data Types | Mixed Allowed | Same Type Only |
| Memory | Managed Automatically | Lower-level memory management |
| Built-in Operations | Many Available | More Limited |
| Resizable | Yes | No |
| Ease of Use | Very Easy | More Complex |
| Indexing | Yes | Yes |
| Slicing | Supported | Not directly supported |


## Example: C++ Array

```cpp
int numbers[5] = {10, 20, 30, 40, 50};
```

### Limitations

- Fixed size.
- Same data type.
- Cannot easily add or remove elements.


## Equivalent Python List

````python
numbers = [10, 20, 30, 40, 50]

numbers.append(60)

print(numbers)
````

### Advantages

- Dynamic resizing.
- Rich built-in methods.
- Easier syntax.
- Can store mixed types.


# Summary

- A **List** is an ordered and mutable collection in Python.
- Lists support indexing, slicing, insertion, deletion, and sorting.
- Python lists are dynamic and more flexible than C/C++ arrays.
- Lists can store mixed data types and nested collections.
- Common methods include `append()`, `insert()`, `remove()`, `pop()`, `sort()`, and `reverse()`.
- Lists are among the most frequently used data structures in Python, AI, Data Science, and Software Development.
