# Packages in Python

When teaching packages, it is useful to show the hierarchical structure first.

## Example Directory Structure

```text

university/
│
├── __init__.py
│
├── faculty/
│   ├── __init__.py
│   ├── teachers.py
│   └── staff.py
│
└── students/
    ├── __init__.py
    └── records.py

```

Here:

- **university** → Package

- **faculty** and **students** → Sub-packages

- **teachers.py**, **staff.py**, **records.py** → Modules

---

## Example Module

### `teachers.py`

````python
def display_teacher():

    print("Dr. Ahmad")
````

---

# Method 1: Import the Module from a Sub-package

````python
import university.faculty.teachers

university.faculty.teachers.display_teacher()
````

**Output**

```text

Dr. Ahmad

```

---

# Method 2: Import the Module Directly

````python
from university.faculty import teachers

teachers.display_teacher()
````

**Output**

```text

Dr. Ahmad

```

---

# Method 3: Import the Function Directly

````python
from university.faculty.teachers import display_teacher

display_teacher()
````

**Output**

```text

Dr. Ahmad

```

---

# Method 4: Use an Alias

````python
import university.faculty.teachers as t

t.display_teacher()
````

**Output**

```text

Dr. Ahmad

```

# Accessing a Module in a Nested Sub-package

Suppose the structure becomes:

```text

university/
│
└── faculty/
    │
    └── engineering/
        │
        ├── __init__.py
        └── computer_science.py

```

### `computer_science.py`

````python
def course():

    print("Data Structures")
````

---

## Importing from a Nested Sub-package

````python
import university.faculty.engineering.computer_science

university.faculty.engineering.computer_science.course()
````

or

````python
from university.faculty.engineering import computer_science

computer_science.course()
````

or

````python
from university.faculty.engineering.computer_science import course

course()
````

# Relative Imports (Inside Packages)

Suppose `staff.py` wants to use a function from `teachers.py`.

### `teachers.py`

````python
def display_teacher():

    print("Dr. Ahmad")
````

### `staff.py`

````python
from .teachers import display_teacher

display_teacher()
````

The dot (`.`) means:

```text

Current Package

```

Similarly,

````python
from ..students.records import show_record
````

means:

```text

Go one level up,

then enter students package,

then import records module.

```

# Rule for Students

A good memory aid is:

```text

Package
   └── Sub-package
          └── Module
                 └── Function/Class

```

and the import statement generally follows the same path:

````python
from package.subpackage.module import function
````

For example:

````python
from university.faculty.teachers import display_teacher
````

which reads as:

```text

From package "university"

    go to sub-package "faculty"

        go to module "teachers"

            import function "display_teacher"

```

This analogy helps students quickly understand how Python navigates through packages, sub-packages, modules, and functions.
