File Handling: Reading and Writing Text Files

## 1. Definition: File Handling

File Handling is the process of <u>creating</u>, <u>opening</u>, <u>reading</u>, <u>writing</u>, <u>updating</u>, and <u>closing files</u> stored on a computer. It allows programs to store data permanently instead of keeping it only in memory during execution.

#### Explanation

- Files provide permanent storage of data.
- Data remains available even after the program terminates.
- Python provides built-in functions for file operations.
- Files can contain text, numbers, images, or other data.
- Common file operations include opening, reading, writing, and closing.
- Python treats files as objects, it means that when you open a file, Python gives you a file object that you can use to interact with that file.

#### Example

```python
file = open("data.txt", "r")
print(file.read())
file.close()
```

## 2. Definition: Reading Text Files

Reading a Text File means retrieving data stored inside a text file and loading it into a Python program.

#### Explanation

- The file must usually be opened in read mode ("r").
- Reading does not modify the file contents.
- Python provides several methods for reading data.
- Entire files or individual lines can be read.
- Useful for loading configuration files, logs, and datasets.
- Attempting to read a non-existing file causes an error.

## 3. Reading Using read()

#### Explanation

- Reads the entire file as one string.
- Can also read a specific number of characters.
- Cursor moves to the end after reading.
- Suitable for small files.
- Easy to use but may consume more memory for large files.

#### Example

```python
file = open("students.txt", "r")
content = file.read()
print(content)
file.close()
```

### Reading Limited Characters

```python
file = open("students.txt", "r")
data = file.read(5)
print(data)
file.close()
```

## 4. Reading Using readline()

#### Explanation

- Reads one line at a time.
- Useful for processing large files.
- Cursor moves to the next line automatically.
- Returns an empty string when the file ends.
- Helps reduce memory usage.

#### Example

```python
file = open("students.txt", "r")

print(file.readline())
print(file.readline())

file.close()
```

## 5. Reading Using readlines()

#### Explanation

- Reads all lines and stores them in a list.
- Each line becomes one list element.
- Useful when line-by-line access is needed.
- Easier for iteration.
- Consumes memory for large files.

#### Example

```python
file = open("students.txt", "r")

lines = file.readlines()
print(lines)

file.close()
```

## 6. Definition: Writing Text Files

Writing a Text File means storing new data into a file.

#### Explanation

- Open the file in write mode ("w").
- Creates a new file if it does not exist.
- Existing contents are overwritten.
- Useful for saving reports, logs, and user input.
- Data must be written as strings.

#### Example

```python
file = open("notes.txt", "w")

file.write("Python is easy to learn.")

file.close()
```

## 7. Writing Multiple Lines

#### Explanation

- Multiple calls to write() can be used.
- Newline character \\n starts a new line.
- Useful for creating structured text files.
- Data is written sequentially.

#### Example

```python
file = open("students.txt", "w")

file.write("Ali\n")
file.write("Sara")
file.write("Ahmed\n")
file.write("khan")

file.close()
```

## 8. Appending Data to a File

#### Explanation

- Uses append mode ("a").
- New data is added at the end.
- Existing contents remain unchanged.
- Useful for logs and records.
- Creates file if it does not exist.

#### Example

```python
file = open("students.txt", "a")

file.write("Bilal\n")

file.close()
```

## Summary

Students should understand:

- What file handling is.
- How to open and close files.
- Reading files using read(), readline(), and readlines().
- Writing data using write().
- Appending data using "a" mode.
- Importance of permanent data storage.
