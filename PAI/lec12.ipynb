# Importing Modules in Python

## Definition

The `import` statement is used to include and use code from another Python module in the current program. It allows code reuse and helps organize large programs into smaller files.

## Key Points

- **Promotes Code Reusability:** Write code once and use it in multiple programs.
- **Improves Organization:** Large projects can be divided into multiple modules.
- **Avoids Code Duplication:** Common functionality can be placed in separate files.
- **Supports Selective Importing:** Specific functions or variables can be imported.
- **Can Use Aliases:** Long module names can be shortened using `as`.
- **Python Searches for Modules:** In the current directory and predefined paths.


## Examples

### Importing an Entire Module

````python
import math

print(math.sqrt(25))
````

### Importing Specific Functions


````python
from math import sqrt

print(sqrt(36))
````

### Using an Alias


````python
import math as m

print(m.pi)
````


# Modules in Python

## Definition

A **module** is a Python file (`.py`) containing <u>variables</u>, <u>functions</u>, and <u>classes</u> that can be imported and used in another Python program.


## Key Points

- A module is simply a Python file.
- Encapsulates related functionality.
- Makes programs easier to maintain.
- Can contain functions, variables, and classes.
- Can be imported into other Python programs.
- Helps divide large applications into manageable parts.


## Example

### File: `calculator.py`

````python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b
````

### Main Program


````python
import calculator

print(calculator.x)
print(calculator.add(5, 3))
print(calculator.multiply(4, 2))
````

# Python `import` vs C/C++ `#include`

### Python `import`

Loads a module as a separate namespace and executes it only once. A namespace is a container that maps names to objects and helps prevent naming conflicts.

### Explaination 
When Python imports a module, it creates a separate place for that module's names (variables, functions, classes, etc.), runs the module's code, and then reuses that already-loaded module if it is imported again.

### C/C++ `#include`

Copies the contents of a header file into the source file before compilation.


## Key Differences

| Feature | Python `import` | C/C++ `#include` |
|----------|-----------------|------------------|
| Purpose | Load modules | Copy header file contents |
| Performed By | Python Interpreter | Preprocessor |
| Code Duplication | No | Textually copied |
| Namespace Support | Yes | No |
| Execution | Module executed once | Header expanded every time |
| Typical Usage | `import math` | `#include <iostream>` |


## Examples

### Python

````python
import math

print(math.sqrt(16))
````

### C++

```cpp
#include <iostream>

using namespace std;

int main()
{
    cout << "Hello";
}
```


## Analogy

Think of:

```cpp
#include
```

as **copy-paste the file here**

while

```python
import
```

means **load and use the module when needed**.

# Built-in Modules in Python

## Definition

Built-in modules are modules that come pre-installed with Python and can be used immediately <u>without any additional installation</u> such as `math`.


## Key Points

- Available by default with Python.
- No `pip` (Package Manager) installation required.
- Provide commonly needed functionality.
- Developed and maintained as part of Python.
- Improve programmer productivity.
- Used extensively in real-world applications.



## Common Built-in Modules

| Module | Purpose |
|----------|---------|
| `math` | Mathematical functions |
| `random` | Random number generation |
| `os` | Operating system interaction |
| `sys` | Python interpreter information |
| `datetime` | Date and time handling |
| `statistics` | Statistical calculations |
| `json` | JSON data processing |



### Example: `math` Module


````python
import math

print(math.factorial(5))
````


### Example: `random` Module


````python
import random

print(random.randint(1, 10))
````


### Example: `datetime` Module


````python
from datetime import date

print(date.today())
````

````python
# alternate code
import datetime

print(datetime.date.today())
````

# Quick Summary

| Term | Description |
|------|-------------|
| **Module** | A single Python file containing reusable code |
| **Package** | A directory containing multiple related modules |
| **Import** | Mechanism used to load modules/packages |
| **Built-in Module** | Module already included with Python |
| **`#include`** | C/C++ preprocessor directive that copies header contents |
