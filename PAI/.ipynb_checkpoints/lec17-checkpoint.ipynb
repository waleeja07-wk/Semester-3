### Flow of Control in try–except

#### Explanation

- Statements in `try` run one by one.
- As soon as an exception occurs, the **rest of the try block is skipped**.
- Control jumps to the matching `except` block.
- After the `except` block, the program continues with the next statement after the whole try–except structure.

#### Example

```python
print("Program started")

try:
    print("Step 1")
    x = 10 / 0
    print("Step 2")        # skipped
except ZeroDivisionError:
    print("Handled division by zero")

print("Program continues")
```

### Exceptions Inside Functions

If an exception occurs inside a function and is not handled there, it travels back to the place where the function was called. It can be handled by the caller.

```python
def get_percentage(obtained, total):
    return obtained / total * 100

try:
    print(get_percentage(45, 0))
except ZeroDivisionError:
    print("Total marks cannot be zero.")
```

## 4. The else Clause

#### Definition

The `else` block runs **only if no exception occurred** in the `try` block.

#### Explanation

- Written after all `except` blocks.
- Keeps the `try` block small; only the risky statement goes inside `try`.
- Code that should run only on success goes inside `else`.
- If an exception occurs, `else` is skipped.

#### Syntax

```python
try:
    # risky code
except ExceptionType:
    # runs if an exception occurs
else:
    # runs if no exception occurs
```

#### Example

```python
try:
    num = int(input("Enter a number: "))
except ValueError:
    print("Invalid input.")
else:
    print("Double of the number is", num * 2)
```

## 5. Practical Examples

### Example 1: Keep Asking Until Input Is Valid

A `while True` loop combined with try–except keeps asking the user until a correct value is entered.

```python
while True:
    try:
        age = int(input("Enter your age: "))
        break
    except ValueError:
        print("Please enter digits only. Try again.")

print("Your age is", age)
```

### Example 2: Safe List Access

```python
fruits = ["apple", "banana", "mango"]

try:
    index = int(input("Enter index (0-2): "))
    print("Fruit:", fruits[index])
except IndexError:
    print("Index out of range.")
```

### Example 3: Dictionary Lookup

```python
prices = {"pen": 50, "notebook": 120, "bag": 1500}

item = input("Enter item name: ")

try:
    print("Price of", item, "is Rs.", prices[item])
except KeyError:
    print(item, "is not available.")
```

### Example 4: Converting Mixed Data

Convert a list of strings to integers, skipping values that are not valid numbers.

```python
data = ["10", "25", "abc", "40", "4.5", "7"]
numbers = []

for item in data:
    try:
        numbers.append(int(item))
    except ValueError:
        print("Skipping invalid value:", item)

print("Valid numbers:", numbers)
print("Sum:", sum(numbers))
```

### Example 5: Simple Safe Calculator

```python
try:
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))
    op = input("Enter operator (+, -, *, /): ")

    if op == "+":
        print("Result =", a + b)
    elif op == "-":
        print("Result =", a - b)
    elif op == "*":
        print("Result =", a * b)
    elif op == "/":
        print("Result =", a / b)
    else:
        print("Unknown operator")

except ValueError:
    print("Please enter numeric values only.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

## Common Mistakes


| Mistake | Why it is a problem | Better approach |
|---|---|---|
| Using bare `except:` everywhere | Hides unexpected bugs such as typos | Catch the specific exception, e.g. `except ValueError:` |
| Putting too much code inside `try` | Hard to know which line caused the error | Keep only the risky statements in `try`; use `else` for the rest |
| Empty handler (`except: pass`) | The error is silently ignored | Show a message or take corrective action |
| Expecting try/except to catch `SyntaxError` in the same file | Syntax errors stop the program before it runs | Fix the code |
| Using try/except instead of a simple check | Makes logic harder to follow | Use `if` when the condition is easy to test |


## 2. The finally Block

#### Definition

The finally block contains code that executes whether an exception occurs or not.

#### Explanation

- Executes after try and except blocks.
- Runs even if an exception occurs.
- Often used for cleanup tasks.
- Commonly used for closing files.
- Ensures important code always executes.
- Improves program reliability.

#### Syntax

```python
try:
    # risky code

except:
    # handling code

finally:
    # always executes
```

#### Example

```python
try:
    num = int(input("Enter number: "))
    result = 10 / num
    print(result)
except ZeroDivisionError:
    print("Division by zero")
finally:
    print("Program ended")
```

```python
try:
    num = int(input("Enter number: "))
    result = 10 / num
    print(result)
except ZeroDivisionError:
    print("Division by zero")
print("Program ended")              # without finally
```

## 3. The raise Statement

#### Definition

The raise statement is used to generate an exception intentionally.

#### Explanation

- Allows programmers to create exceptions manually.
- Useful for validating input.
- Stops normal execution when conditions are violated.
- Can raise built-in exceptions.
- Helps enforce program rules.
- Improves code safety and correctness.

#### Syntax

```python
raise ExceptionType("Message")
```

#### Example

```python
age = int(input("Enter age: "))

if age < 0:
    raise ValueError("Age cannot be negative")

print("Age =", age)
```

### Example: Custom Validation Using raise

```python
marks = int(input("Enter marks: "))

if marks < 0 or marks > 100:
    raise ValueError("Marks must be between 0 and 100")

print("Valid Marks")
```

## 4. Best Practices in Exception Handling

#### Definition

Best practices are recommended techniques for writing effective, readable, and reliable exception-handling code.

#### Explanation

- Catch specific exceptions whenever possible.
- Avoid using a bare except: unnecessarily.
- Keep try blocks small and focused.
- Use meaningful error messages.
- Use finally for cleanup operations.
- Raise exceptions when invalid conditions occur.

### Poor Practice

```python
try:
    num = int(input())
    result = 100 / num
except:
    print("Error")
```

### Better Practice

```python
try:
    num = int(input())
    result = 100 / num

except ValueError:
    print("Please enter a valid integer.")

except ZeroDivisionError:
    print("Number cannot be zero.")
```

### Comprehensive Example

```python
try:
    age = int(input("Enter age: "))

    if age < 0:
        raise ValueError("Age cannot be negative")

    print("Age =", age)

except ValueError as e:
    print("Error:", e)

finally:
    print("Thank you for using the program.")
```

### Same logic, but using an inappropriate exception type

```python
try:
    age = int(input("Enter age: "))

    if age < 0:
        raise NameError("Age cannot be negative")

    print("Age =", age)

except NameError as e:
    print("Error:", e)

finally:
    print("Thank you for using the program.")
```

## Summary

- Multiple except blocks handle different exceptions separately.
- finally executes regardless of exceptions.
- raise generates exceptions intentionally.
- Specific exception handling is preferred over generic handling.
- Good exception handling improves robustness, readability, and maintainability of Python programs.
- Exception handling is essential for building reliable real-world software.
