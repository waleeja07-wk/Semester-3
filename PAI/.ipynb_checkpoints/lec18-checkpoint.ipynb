## 1. Classes

#### Definition

A Class is a blueprint or template used to create objects. It defines the properties (attributes) and behaviors (methods) that objects will have.

#### Explanation

- A class acts as a user-defined data type.
- It groups data and functions together.
- Objects created from the same class share the same structure.
- Python uses the class keyword to define a class.
- Unlike C++, Python does not require header files or access specifiers for basic class creation.

#### Real-Life Example

- A **house map** (blueprint) is a class. You cannot live in a map, but you can build many houses from one map.
- Class names start with a **Capital letter** by convention, e.g. `Student`, `BankAccount`.
- `pass` means "do nothing for now". We use it when the class is empty.

#### Example

````python
class Student:
    pass

s1 = Student()
print(type(s1))
````

#### Example 2

````python
class Car:
    pass

class Book:
    pass

c = Car()
b = Book()

print(type(c))
print(type(b))
````

**Note:** Each class is a new data type, just like `int` or `str`.

## 2. Objects

#### Definition

An Object is an instance of a class. It represents a real-world entity and contains actual data.

#### Explanation

- Objects are created from classes.
- Each object has its own data.
- Multiple objects can be created from one class.
- Objects interact through methods.
- Memory is allocated when an object is created.

#### Real-Life Example

- Class = cookie cutter, Object = cookie.
- One cutter can make many cookies. Each cookie is separate.
- To create an object, write the class name with brackets: `Student()`

#### Example

````python
class Student:
    pass

s1 = Student()
s2 = Student()

print(s1)
print(s2)
````

**Note:** The numbers after `at` are memory addresses. They are different, so `s1` and `s2` are two different objects. You will see different numbers on your computer.

#### Example 2

````python
class Student:
    pass

s1 = Student()
s2 = Student()
s3 = s1          # s3 is NOT a new object, it points to s1

print(s1 == s2)
print(s1 == s3)
print(isinstance(s1, Student))
````

#### Explanation

- `s1 == s2` is `False` because they are two different objects.
- `s3 = s1` does not make a new object. Both names point to the same object.
- `isinstance(s1, Student)` checks "Is `s1` made from the `Student` class?"

## 3. Constructor (\_\_init\_\_)

#### Definition

The \_\_init\_\_ method is a special method automatically called when an object is created. It initializes object attributes.

General Syntax
```python
class ClassName:
    def __init__(self, parameters):
        self.attribute = parameters
```

#### Explanation

- Called automatically during object creation.
- Used to assign initial values.
- Similar to constructors in C++.
- Every object can have different initial values.
- The first parameter is always self.

#### Real-Life Example

When a baby is born, it gets a name on day one. `__init__` does the same: it gives starting values to an object the moment it is created.

- It has **two underscores** before and after `init`.
- We never call `__init__` ourselves. Python calls it when we write `Student(...)`.

#### Example

````python
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Ali")

print(s1.name)
````

#### Example 2: \_\_init\_\_ Runs Automatically

````python
class Phone:
    def __init__(self, brand):
        print("A new phone is being created...")
        self.brand = brand

p1 = Phone("Samsung")
p2 = Phone("Apple")

print(p1.brand)
print(p2.brand)
````

**Note:** We never wrote `p1.__init__()`. The message printed two times because we created two objects.

#### Example 3: Default Values

````python
class Student:
    def __init__(self, name, city="Lahore"):
        self.name = name
        self.city = city

s1 = Student("Ali")
s2 = Student("Sara", "Karachi")

print(s1.name, "-", s1.city)
print(s2.name, "-", s2.city)
````

**Note:** If no city is given, the default value `"Lahore"` is used. This is like default arguments in a C++ constructor.

## 4. The self Keyword


#### Definition

self refers to the current object of the class.

#### Explanation

- Similar to `this` pointer in C++.
- Used to access attributes and methods.
- Must be the first parameter in instance methods.
- Python passes it automatically.
- Not a keyword; just a naming convention.

#### Real-Life Example

When a teacher says "open **your** notebook", every student opens **their own** notebook. `self` works the same way: it means "**this** object that called the method".

#### Example

````python
class Test:
    def show(self):
        print("Current object:", self)

t = Test()
t.show()
````

#### Example 2: self Is the Object Itself

````python
class Test:
    def show(self):
        print("Address of self:", id(self))

t = Test()
print("Address of t:   ", id(t))
t.show()
````

**Note:** Both addresses are the same. When we write `t.show()`, Python actually runs `Test.show(t)`, so `t` goes into `self`. `id()` is a built-in Python function that returns a unique identity number for an object during its lifetime.

#### Example 3: self Keeps Each Object's Data Separate

````python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        print(self.name, "says Woof!")

d1 = Dog("Tommy")
d2 = Dog("Rocky")

d1.bark()     # self = d1
d2.bark()     # self = d2
````

## 5. Methods

#### Definition

Methods are functions defined inside a class that describe object behavior.

#### Explanation

- Defined using def.
- Operate on object data.
- Can receive parameters.
- Can return values.
- Called through objects.

#### Function vs Method

| Function | Method |
|---|---|
| Written outside a class | Written inside a class |
| Called by name: `greet()` | Called with an object: `s.greet()` |
| No `self` | First parameter is `self` |

#### Example

````python
class Student:
    def greet(self):
        print("Welcome Student")

s = Student()
s.greet()
````

#### Example 2: Method with Parameters

````python
class Calculator:
    def add(self, a, b):
        print(a, "+", b, "=", a + b)

    def multiply(self, a, b):
        print(a, "x", b, "=", a * b)

calc = Calculator()
calc.add(5, 3)
calc.multiply(4, 6)
````

**Note:** We pass only 2 values, but the method has 3 parameters (`self, a, b`). Python fills `self` automatically.

#### Example 3: Method That Returns a Value

````python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

r = Rectangle(5, 3)
print("Area:", r.area())
print("Perimeter:", r.perimeter())
````

#### Example 4: Method That Changes Object Data

````python
class Counter:
    def __init__(self):
        self.count = 0

    def increase(self):
        self.count = self.count + 1

c = Counter()
c.increase()
c.increase()
c.increase()

print("Count:", c.count)
````

**Note:** The object remembers `count` between method calls.

## 6. Attributes

#### Definition

Attributes are variables belonging to an object or class.

#### Explanation

- Store object information.
- Accessed using dot notation.
- Created inside \_\_init\_\_ or class body.
- Each object may have different values.
- Similar to data members in C++.

#### Two Types of Attributes

| Instance Attribute | Class Attribute |
|---|---|
| Written with `self.` inside `__init__` | Written directly inside the class |
| Each object has its own copy | Shared by all objects |
| Example: student name | Example: university name |
| Like normal data members in C++ | Like `static` data members in C++ |

#### Example

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s = Student("Sara", 20)

print(s.name)
print(s.age)
````

#### Example 2: Changing Attribute Values

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s = Student("Sara", 20)
print("Before:", s.age)

s.age = 21
print("After:", s.age)
````

#### Example 3: Instance vs Class Attribute

````python
class Student:
    university = "FAST NUCES"          # class attribute (shared)

    def __init__(self, name):
        self.name = name                # instance attribute (own copy)

s1 = Student("Ali")
s2 = Student("Sara")

print(s1.name, "-", s1.university)
print(s2.name, "-", s2.university)
````

#### Example 4: Counting Objects

````python
class Student:
    total = 0

    def __init__(self, name):
        self.name = name
        Student.total = Student.total + 1

s1 = Student("Ali")
s2 = Student("Sara")
s3 = Student("Ahmed")

print("Total students:", Student.total)
````

### Complete Example

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

s1 = Student("Ali", 21)
s1.display()
````

### Complete Example 2: Bank Account

This example uses everything from today: class, `__init__`, attributes, methods and `self`.

#### Example

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        self.balance = self.balance + amount
        print(amount, "deposited.")
    def withdraw(self, amount):
        if amount > self.balance:
            print("Sorry! Not enough balance.")
        else:
            self.balance = self.balance - amount
            print(amount, "withdrawn.")
    def show_balance(self):
        print(self.owner, "has Rs.", self.balance)

acc1 = BankAccount("Ali", 1000)
acc2 = BankAccount("Sara")
acc1.deposit(500)
acc1.withdraw(2000)
acc1.withdraw(300)
acc2.deposit(700)
acc1.show_balance()
acc2.show_balance()
````

#### Explanation of previous code

- `acc1` and `acc2` are separate. Money in one does not affect the other.
- `acc1.withdraw(2000)` failed because Ali had only 1500.
- `acc2` started with the default balance `0`.

### Task
In the above code, create a class attribute that counts the number of instances created.

````python
class BankAccount:
    account_count = 0  # Class attribute
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        BankAccount.account_count += 1
    def deposit(self, amount):
        self.balance = self.balance + amount
        print(amount, "deposited.")
    def withdraw(self, amount):
        if amount > self.balance:
            print("Sorry! Not enough balance.")
        else:
            self.balance = self.balance - amount
            print(amount, "withdrawn.")
    def show_balance(self):
        print(self.owner, "has Rs.", self.balance)

acc1 = BankAccount("Ali", 1000)
acc2 = BankAccount("Sara")
acc1.deposit(500)
acc1.withdraw(2000)
acc1.withdraw(300)
acc2.deposit(700)
acc1.show_balance()
acc2.show_balance()
print("Total accounts created:", BankAccount.account_count)
````

### Python OOP vs C++

| Feature | Python | C++ |
|---|---|---|
| Class Definition | class | class |
| Constructor | init() | ClassName() |
| Object Creation | Student() | Student s; |
| Access Specifiers | Optional Convention | public/private/protected |
| this/self | self | this |
| Semicolon | Not Required | Required |
| Memory Management | Automatic | Manual/Automatic |

#### Same Class in C++ and Python

**C++**

```cpp
class Student {
public:
    string name;
    int age;

    Student(string n, int a) {
        name = n;
        age = a;
    }

    void display() {
        cout << "Name: " << name << endl;
        cout << "Age: " << age << endl;
    }
};

Student s1("Ali", 21);
s1.display();
```

**Python**

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

s1 = Student("Ali", 21)
s1.display()
```

### Explanation
- In C++, data members must be declared with their type first. In Python, they are created by `self.name = ...`.
- C++ uses `this` silently. Python writes `self` openly in every method.
- C++ uses `{ }` and `;`. Python uses `:` and indentation.

## Summary

- \_\_init\_\_ initializes object data.
- self refers to the current object.
- Methods define behavior.
- Attributes store data.
