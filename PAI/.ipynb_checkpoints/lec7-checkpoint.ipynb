# 1. for Loop

A **for loop** is used to iterate over a sequence (such as a list, tuple, string, dictionary, or range) and execute a block of code repeatedly for each item.



## Explanation

- Used when the number of iterations is known or can be determined.
- Iterates through collections and sequences.
- Simpler and safer than a `while` loop for fixed iterations.
- Commonly used for processing data structures.
- Similar to range-based loops in modern C++.


## Syntax

```python
for variable in sequence:
    statements
```

## Example 1: Using `range()`
The `range()` function generates a sequence of numbers within a specified range, commonly used for looping a specific number of times.


````python
for i in range(5):
    print(i)
````

## Example 2: Iterating over a list


````python
fruits = ["Apple", "Banana", "Orange"]

for fruit in fruits:
    print(fruit)
````

# 2. while Loop

A **while loop** repeatedly executes a block of code as long as a specified condition remains `True`.

## Explanation

- Used when the number of iterations is not known beforehand.
- Condition is checked before each iteration.
- Can potentially run forever if the condition never becomes `False`.
- Useful for user input and event-driven programs.
- Similar to the `while` loop in C/C++.

## Syntax

```python
while condition:
    statements
```

## Example

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

# 3. `break` Statement

The **break** statement immediately terminates the nearest loop.



## Explanation

- Exits the loop before completion.
- Often used when a desired condition is met.
- Works in both `for` and `while` loops.
- Improves efficiency by avoiding unnecessary iterations.
- Similar to `break` in C/C++.


## Example


````python
for i in range(10):
    if i == 5:
        break
    print(i)
````

# 4. continue Statement

The **continue** statement skips the remaining code in the current iteration and moves to the next iteration.



## Explanation

- Does not terminate the loop.
- Skips only the current iteration.
- Useful for filtering unwanted values.
- Works in both `for` and `while` loops.
- Similar to `continue` in C/C++.


## Example


````python
for i in range(6):
    if i == 3:
        continue
    print(i)
````

# 5. Nested Loops

A **nested loop** is a loop placed inside another loop.


## Explanation

- Outer loop controls the number of repetitions.
- Inner loop completes all iterations for each outer loop iteration.
- Useful for matrices, tables, and patterns.
- Can be nested multiple levels deep.
- Time complexity increases significantly with nesting.


## Example

````python
for i in range(3):
    for j in range(2):
        print(i, j)
````

# Variants of for Loop

## Variant 1: Loop Through String

````python
for ch in "Python":
    print(ch)
````

## Variant 2: Loop Through List

````python
numbers = [10, 20, 30]

for num in numbers:
    print(num)
````

````python
# List Comprehension (Compact form of for)
squares = [x*x for x in range(5)]
print(squares)

print("------Equivalent Detailed/shortcut form of above -------")
lst = []
for x in range(5):
    lst.append(x*x)
print(lst)
````

## Variant 3: Using `enumerate()`
`enumerate()` — Use `enumerate()` when you need both the index and the value while looping through a sequence.

````python
colors = ["Red", "Green", "Blue"]

for index, color in enumerate(colors):
    print(index, color)
````

````python
fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
    
print('===== 1 ====')
for fruit in enumerate(fruits):
    print(fruit)
    
print('===== 2 ====')
for fruit in enumerate(fruits):
    print(index, fruit)
````

## Variant 4: Using `zip()`
`zip()` — Use `zip()` to iterate over two or more sequences simultaneously, pairing their corresponding elements together.

````python
names = ["Ali", "Ahmed", "Sara"]
marks = [80, 90, 95]

for name, mark in zip(names, marks):
    print(name, mark)
````

## Variant 5: Dictionary Iteration

````python
student = {
    "name": "Ali",
    "age": 20
}

for key, value in student.items():
    print(key, value)
````

# Variants of while Loop

## Variant 1: Counter-Controlled Loop


````python
i = 1

while i <= 5:
    print(i)
    i += 1
````

## Variant 2: Sentinel-Controlled Loop

A **sentinel-controlled loop** continues until a special value is entered.



````python
data = ""

while data != "exit":
    data = input("Enter command: ")
````

## Variant 3: Infinite Loop

```python
while True:
    print("Running...")
```


An infinite loop must usually contain a `break` statement when you want a controlled exit.

````python
while True:
    choice = input("Enter q to quit: ")

    if choice == "q":
        break
````

## Variant 4: Validation Loop

A validation loop repeatedly asks for input until the input satisfies a required condition.

````python
number = -1

while number < 0:
    number = int(input("Enter positive number: "))
````

## Variant 5: Menu-Driven Program

````python
choice = 0

while choice != 3:

    print("1. Add")
    print("2. Delete")
    print("3. Exit")

    choice = int(input("Choice: "))
````

# Loop Control Statements Summary

| Statement | Purpose |
|-----------|---------|
| `break` | Terminates the loop immediately |
| `continue` | Skips the current iteration |
| `pass` | Placeholder; does nothing |


## Example of `pass`
The `pass` statement is used when you want to leave a block of code empty without causing an error.

````python
for i in range(5):
    if i == 2:
        pass
    print(i)
````

# Comparison: for vs while

| Feature | `for` Loop | `while` Loop |
|---------|------------|--------------|
| Number of iterations known | Yes | Usually No |
| Condition required | No | Yes |
| Risk of infinite loop | Low | High |
| Best for sequences | Yes | Not ideal |
| Best for user input | Limited | Excellent |
| Similar to C++ | Range-based `for` | `while` |


# Practical Examples

## Multiplication Table



````python
num = 5
for i in range(1, 11):
    print(num, "x", i, "=", num * i)
````

## Sum of Numbers

````python
total = 0

for i in range(1, 6):
    total += i

print(total)
````

## Factorial Using `while`

````python
n = 5
fact = 1

while n > 0:
    fact *= n
    n -= 1

print(fact)
````

# Key Takeaways

- Use **`for`** when iterating over sequences or when the number of iterations is known.
- Use **`while`** when repetition depends on a condition.
- **`break`** exits a loop immediately.
- **`continue`** skips the current iteration.
- Nested loops are useful for tables, matrices, and patterns.
- Python's `for` loop is more powerful than C/C++ because it can directly iterate over lists, strings, tuples, sets, dictionaries, files, and many other iterable objects.
- Always ensure that a `while` loop eventually becomes `False` or contains a `break` statement to avoid unintended infinite loops.
