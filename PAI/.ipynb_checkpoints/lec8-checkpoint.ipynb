## 1. Defining and Calling Functions 

A **function** is a reusable block of code that performs a specific task. Functions help reduce code duplication, improve readability, and make programs easier to maintain.

### Why Use Functions? 
- Break large programs into smaller manageable parts. 
- Avoid repeating the same code multiple times. 
- Improve code readability and organization. 
- Simplify debugging and maintenance. 
- Promote code reusability.

### Basic Syntax 
```python 
    def function_name(): 
        # Function body 
        print("Hello World") 
```

### Calling a Function

````python
def greet(): 
    print("Welcome to Python") 

greet()
````

### Function Execution Flow 

1. Function is defined using `def`.
2. Python stores the function in memory. 
3. Function code does not execute until it is called. 
4. When called, Python executes the statements inside the function.

# 2. Parameters and Arguments 

A **parameter** is a variable listed in the function definition.<br> 
An **argument** is the actual value passed to the function during the call.

### Syntax 
```python 
def function_name(parameter): 
    # code 

function_name(argument) 
```

### Example

````python
def greet(name): 
    print("Hello", name) 
    
greet("Ali")
````

- `name` → Parameter 
- `"Ali"` → Argument

# 3. Return Values 

A function can send a result back to the caller using the `return` statement.

### Syntax 
```python 
def function_name(): 
    return value 
```

### Example

````python
def add(): 
    return 10 + 20 

result = add() 
print(result)
````

### Why Return Values? 

- Store results in variables. 
- Use results in calculations. 
- Pass results to other functions.

# Functions with Different Data Types

## A. Integer Parameters and Return Values 

### Example 1: Addition 

````python
def add(a, b): 
    return a + b 

result = add(5, 3) 
print(result)
````

### Example 2: Multiplication

````python
def multiply(x, y): 
    return x * y 

print(multiply(4, 6))
````

## B. Float Parameters and Return Values

### Example 1: Calculate Area of Circle

````python
def area_circle(radius): 
    return 3.14159 * radius * radius 

print(area_circle(2.5))
````

### Example 2: Average

````python
def average(a, b): 
    return (a + b) / 2 

print(average(4.5, 7.5))
````

## C. String Parameters and Return Values 
### Example 1: Greeting

````python
def greet(name): 
    return "Hello " + name 

print(greet("Umer"))
````

### Example 2: Convert to Uppercase

````python
def to_upper(text): 
    return text.upper() 

print(to_upper("python"))
````

## D. Boolean Parameters and Return Values 
### Example 1: Voting Eligibility

````python
def can_vote(age): 
    return age >= 18 

print(can_vote(20)) 
print(can_vote(15))
````

### Example 2: Check Positive Number

````python
def is_positive(num): 
    return num > 0 

print(is_positive(10)) 
print(is_positive(-5))
````

## E. List Parameters and Return Values

### Example 1: Sum of List

````python
def total(numbers): 
    return sum(numbers) 

print(total([10, 20, 30]))
````

### Example 2: Return Squared List

````python
def square_list(nums):
    result = []         # creating an empty list
    
    for n in nums:
        result.append(n * n)
    
    return result

numbers = [1, 2, 3, 4]
print(square_list(numbers))
````

## F. Tuple Parameters and Return Values 
### Example 1: Access Tuple Elements

````python
def first_item(data): 
    return data[0] 

print(first_item((100, 200, 300)))
````

### Example 2: Return Multiple Values

````python
def student(): 
    return ("Ali", 21, "CS") 

print(student())
````

## G. Dictionary Parameters and Return Values 
### Example 1: Access Dictionary Value

````python
def get_name(student): 
    return student["name"] 

data = {"name": "Ahmed", "age": 20} 

print(get_name(data))
````

### Example 2: Return Dictionary

````python
def create_student(): 
    return { "name": "Sara", "age": 22 } 

print(create_student())
````

## H. Mixed Data Types 

### Example

````python
def student_info(name, age, cgpa): 
    return f"{name} is {age} years old and has CGPA {cgpa}" 

print(student_info("Ali", 21, 3.7))
````

# Function Returning Multiple Values 
Python can return multiple values separated by commas.

````python
def calculate(a, b): 
    return a + b, a - b, a * b 

sum_val, diff_val, prod_val = calculate(10, 5) 

print(sum_val) 
print(diff_val) 
print(prod_val)
````

# Function Without Return Statement

````python
def show_message(): 
    print("Python Function") 

result = show_message() 
print(result)
````

**Explanation:** 
- Functions without `return` automatically return `None`.

| Data Type Declaration | Required | Not Required | 
| Return Statement | Optional (depending on type) | Optional |

# Key Takeaways 

- Functions are reusable blocks of code. 
- Functions are defined using the `def` keyword. 
- Parameters receive values from the caller. 
- Arguments are actual values passed during function calls. 
- `return` sends results back to the caller. 
- Functions can work with integers, floats, strings, booleans, lists, tuples, dictionaries, and mixed data types. 
- Python functions can return multiple values. 
- Functions without `return` automatically return `None`.

# Task 1: Factorial Calculator

### Problem

Write a function named `factorial()` that accepts an integer `n` and returns the factorial of that number.

The factorial of a number is calculated as:

$$
n! = n \times (n-1) \times (n-2) \times \cdots \times 1
$$

For example:

```text
5! = 5 × 4 × 3 × 2 × 1 = 120
```

````python
# Solution
def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result = result * i

    return result


print(factorial(5))
print(factorial(7))
print(factorial(0))
````

# Task 2: Computing `x` power `n`

### Problem

Write a function named `power()` that accepts a number `x` and a power `n` and returns `x` raised to the power `n` (Without using built-in functions or operators.).

For example:

```text
power(3, 3) → 27
power(5, 5) → 125
```

### Expected Output

```text
27
3125
```


````python
def power(x, n):
    result = 1
    
    for i in range(1, n+1):
        result = result * x
    
    return result

print(power(3, 3))
print(power(-4, 5))
print(power(5, 5))
````

# Task 3: Even or Odd Checker

### Problem

Write a function named `check_even_odd()` that accepts an integer and returns:

- `"Even"` if the number is even.
- `"Odd"` if the number is odd.


### Expected Output

```text
Even
Odd
Even
```

````python
def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"


print(check_even_odd(10))
print(check_even_odd(7))
print(check_even_odd(24))
````
