# 1. Scope in Python


**Scope** refers to the region of a program where a variable can be accessed or used. Python determines the visibility and lifetime of variables using scope rules.


### Types of Scope

1. **Local Scope**
   - Variables defined inside a function.
   - Accessible only within that function.
   - Created when the function is called.
   - Destroyed when the function finishes execution.



````python
def greet():
    message = "Hello"
    print(message)

greet()
# print(message)  # Error: message is not defined
````

2. **Global Scope**
   - Variables defined outside all functions.
   - Accessible throughout the program.
   - Can be read inside functions.
   - The `global` keyword is used when you want to modify/reassign a global variable from inside a function.

````python
name = "Ali"

def show():
    print(name)

show()
````

````python
name = "Ali"

def show():
    name = 'Ahmad'
    print(name)

show()
print(name)
````

3. **Modifying Global Variables**

````python
count = 10

def increase():
    global count
    count += 1

increase()
print(count)
````

````python
count = 10

def increase():
    count = 20       # local variable

    print("Local:", count)

    global count
    
    print(count)
````

Python doesn't allow the same name to be local and global within the same function.

````python
x = "Global"

def outer():
    x = "Enclosing"

    def inner():
        x = "Local"
        print(x)

    inner()
    print(x)
    
outer()
print(x)
````

### Execution Order of the Previous Code

1 → 13 → 4 → 10 → 7 → 8 → 14

4. **LEGB Rule**

Python searches variables in the following order:

| Level | Meaning |
| ----- | ------- |
| L | Local |
| E | Enclosing |
| G | Global |
| B | Built-in |


### What is Built-in?

**Built-in** means names that Python already provides, so you don't have to define them.

For example:

```python
print("Hello")
```

You didn't define `print`, but Python knows it because print is a built-in function.

LEGB Rule
L → E → G → B

Local → Enclosing → Global → Built-in

If Python can't find the name in any of these four scopes, you get a `NameError`.

### Key Points

- Local variables exist only inside functions.
- Global variables can be accessed anywhere.
- Use the `global` keyword to modify global variables.
- Python follows the LEGB rule for variable lookup.
- Proper scope management improves code readability and prevents errors.


# 2. Default Arguments


A **default argument** is a parameter that has a predefined value. If no value is provided during the function call, the default value is used.

### Syntax

```python
def function_name(parameter=default_value):
    pass
```

### Example 1: Single Default Argument

````python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Ahmed")
````

### Example 2: Multiple Default Arguments

````python
def student(name, semester=1):
    print(name, semester)

student("Ali")
student("Sara", 3)
````

### Example 3: Calculator

````python
def power(base, exponent=2):
    return base ** exponent

print(power(5))
print(power(5, 3))
````

### Rules for Default Arguments

````python
# Correct
def func(a, b=10):
    pass

# Incorrect
# def func(a=10, b):
#     pass
````

Non-default parameters must come before default parameters.

### Key Points

- Reduce the need for multiple function definitions.
- Make functions flexible and user-friendly.
- Default values are used when arguments are omitted.
- Non-default parameters must appear first.
- Frequently used in APIs and libraries.


# 3. Lambda Functions


A **lambda function** is a small anonymous function that can have any number of arguments but only one expression. Lambda functions are mainly used for small expressions that produce a value.

### Syntax

```python
lambda arguments: expression
```

Lambda version:

```python
lambda x: x * x
```

Equivalent to:

```python
def square(x):
    return x * x
```

### Example 1: Square Number

````python
square = lambda x: x * x

print(square(5))
````

### Example 2: Addition

````python
add = lambda a, b: a + b

print(add(3, 4))
````

### Example 3: Using `map()`

`map()` applies a function to each element of an iterable (like a list) and returns the transformed values.

```python
map(function, iterable)
``` 
Example: `map(square, numbers)` applies `square()` to every number in `numbers`.

````python
numbers = [1, 2, 3, 4]

squares = list(map(lambda x: x * x, numbers))

print(squares)
print(numbers)
````

````python
# Equivalent Code without Lambda

numbers = [1, 2, 3, 4]

def square(x):
    return x * x

squares = list(map(square, numbers))

print(squares)
print(numbers)
````

### Example 4: Using `filter()`

`filter()` selects elements from an iterable that satisfy a given condition and returns only those elements.

```python
filter(condition, iterable)
``` 

Example: `filter(lambda x: x % 2 == 0, numbers)` keeps only the even numbers.

````python
numbers = [1, 2, 3, 4, 5, 6]

evens = list(filter(lambda x: x % 2 == 0, numbers))

print(evens)
````

````python
# Equivalent Code without Lambda

numbers = [1, 2, 3, 4, 5, 6]

def is_even(x):
    return x % 2 == 0

evens = list(filter(is_even, numbers))

print(evens)
````

### Key Points

- Anonymous functions do not require an explicit function name.
- A lambda function contains a single expression.
- Useful for short, temporary functions.
- Commonly used with `map()`, `filter()`, and `sorted()`.
- Can make simple operations more concise.


# 4. Docstrings


A **Docstring (Documentation String)** is a string written inside a function, class, or module to describe its purpose and usage.

### Syntax

```python
def function_name():
    """Documentation text"""
```

### Example 1: Function Docstring

````python
def add(a, b):
    """
    Returns the sum of two numbers.
    """
    return a + b
````

### Accessing a Docstring

````python
print(add.__doc__)
````

### Example 2: Multi-line Docstring

````python
def divide(a, b):
    """
    Divide two numbers.

    Parameters:
    a -- numerator
    b -- denominator

    Returns:
    Division result
    """
    return a / b

print(divide.__doc__)
````

### Example 3: Class Docstring

````python
class Student:
    """
    Represents a student.
    """

    def __init__(self, name):
        self.name = name

print(Student.__doc__)
````

### Example 4: Module Docstring

````python
"""
This module demonstrates basic mathematical functions.
"""

def square(x):
    return x * x

print(__doc__)
````

A module is a Python file `(.py)` that contains reusable code, such as `functions`, `classes`, and `variables`.

### Key Points

- Improves code readability.
- Helps developers understand code functionality.
- Used by documentation tools.
- Written using triple quotes (`""" """`).
- Can be accessed using the `__doc__` attribute.
