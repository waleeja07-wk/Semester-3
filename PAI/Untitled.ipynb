```python
sentence = input()
count_vowels = 0
count_consonent = 0
for ch in sentence:
    if ch == 'a' or ch == 'e' or ch == 'i' or ch == 'o' or ch == 'u':
        count_vowels = count_vowels + 1
    else:
        count_consonent = count_consonent + 1
print(count_vowels)
print(count_consonent)
```

```python
def classify(n):
    if n%3==0 and n%5==0:
            return "FizzBuzz"
    elif n%3==0:
        return "fizz"
    elif n%5==0:
        return "buzz"
    else:
        return str(n)
for i in range(1, 31):
    print(classify(i))
```

```python
oldlist = [1,2,3,4]
def is_even(x):
    return x % 2 == 0
average = sum(oldlist)/len(oldlist)
newList = list(filter(lambda x: is_even(x) and x > average, oldlist))
print(newList)
```

```python
def is_prime(n):
    if n<2:
        return False
    if n==2  or  n==3:
        return True
    if n%2==0 or n%3==0:
        return False
    i = 5
    while i*i <=n:
        if n%i==0 or n%(i+2)==0:
            return False
        i+=6
    return True
prime = list(filter(is_prime, range(1, 51)))
print(prime)
```
