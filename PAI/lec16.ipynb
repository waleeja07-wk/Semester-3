## 1. Errors vs Exceptions

#### Definition

Errors are problems in a program that prevent it from running correctly.

Exceptions are runtime errors that occur during program execution and can be handled using exception handling mechanisms.

#### Explanation

- Errors and exceptions both indicate abnormal situations in a program.
- Some errors occur before execution (syntax errors), while exceptions occur during execution.
- Errors generally require code correction.
- Exceptions can often be anticipated and handled gracefully.
- Proper exception handling prevents abrupt program termination.
- Python provides many built-in exception classes.

### Common Types of Errors

#### 1. Syntax Error

Occurs when Python code violates language rules.

````python
if 5 > 3
    print("Hello")
````

#### 2. Logical Error

Program runs but produces incorrect results.

````python
length = 5
width = 3

area = length + width   # Wrong formula
print(area)
````

### Errors vs Exceptions: Comparison

| Feature | Errors (Syntax Errors) | Exceptions |
|---|---|---|
| When detected | Before the program runs (while parsing) | While the program is running |
| Cause | Code breaks Python's grammar rules | Valid code meets an unexpected situation |
| Can be handled with try/except? | No | Yes |
| How to fix | Correct the code | Handle it, or prevent the situation |
| Examples | Missing `:`, wrong indentation, unclosed bracket | Dividing by zero, invalid input, missing file |

### Understanding a Traceback

When an exception is not handled, Python stops the program and prints a **traceback**. Reading it tells you exactly what went wrong and where.

````python
def divide(a, b):
    return a / b

print(divide(10, 0))
````

Output:

```
Traceback (most recent call last):
  File "main.py", line 4, in <module>
    print(divide(10, 0))
  File "main.py", line 2, in divide
    return a / b
ZeroDivisionError: division by zero
```

#### How to read it

- Read from the **bottom up**.
- The **last line** gives the exception type (`ZeroDivisionError`) and a message (`division by zero`).
- The lines above show the chain of calls that led to the error.
- The line closest to the bottom is where the exception actually occurred.

## 2. The try Statement

#### Definition

The try block contains code that may generate an exception during execution.

#### Explanation

- Risky code is placed inside the try block.
- Python monitors statements inside the block.
- If no exception occurs, execution continues normally.
- If an exception occurs, Python stops executing the remaining statements in the try block.
- Control transfers to an appropriate except block.
- Using try makes programs more robust.

#### Syntax

````python
try:
    # risky code
````

#### Example

````python
try:
    num = int(input("Enter number: "))
    print("You entered:", num)
except:
    print("Invalid input")
    
print("Execution continues after the try-except block.")
````

````python
num = int(input("Enter number: ")) # same code without try and except
print("You entered:", num)

print("Execution continues after the input and print statements.")
````

## 3. The except Statement

#### Definition

The except block handles exceptions generated inside the corresponding try block.

#### Explanation

- Executes only when an exception occurs.
- Prevents abrupt program termination.
- Can display user-friendly error messages.
- Can handle specific or general exceptions.
- Makes programs more reliable.
- Multiple except blocks can be used for different exceptions.

#### Syntax

````python
try:
    # risky code
except:
    # handling code
````

#### Example

````python
try:
    num = int(input("Enter number: "))
    result = 100 / num
    print(result)
except:
    print("An error occurred")
````

`ValueError` and `ZeroDivisionError` are built-in exception class names in Python, not arbitrary names. You must use their correct names when handling these specific exceptions.

### Catching a Specific Exception

#### Explanation

- A bare `except:` catches **every** exception, including ones you did not expect (even typos like a misspelled variable name).
- This can hide real bugs and give misleading messages.
- It is better to name the exception you expect: `except ValueError:`.
- Exceptions that are not named are **not** caught and will still appear normally.

````python
try:
    print(unknown_variable)

except ValueError:
    print("Invalid input!")

except ZeroDivisionError:
    print("Cannot divide by zero.")
    
except:
    print("couldn't find the error")
````

#### Syntax

````python
try:
    # risky code
except ExceptionType:
    # handling code
````

#### Why a bare except can be misleading

````python
try:
    num = int("25")
    print(nmu)          # typo: should be num
except:
    print("Invalid number entered")
````

Output:

```
Invalid number entered
```

The number was valid; the real problem is a `NameError` caused by a typo, but the bare `except` hides it.

````python
try:
    num = int("25")
    print(nmu)          # typo: should be num
except ValueError:
    print("Invalid number entered")
````

Output:

```
NameError: name 'nmu' is not defined
```

Now the actual bug is visible and can be fixed.

### Accessing the Exception Object (`as`)

#### Explanation

- `except ExceptionType as e:` stores the exception object in the variable `e`.
- `print(e)` shows Python's message describing the error.
- `type(e).__name__` gives the name of the exception class.
- Useful for debugging and logging.

#### Syntax

````python
try:
    # risky code
except ExceptionType as e:
    print(e)
````

#### Example

````python
value = int("12.5")
````

````python
try:
    value = int("12.5") # replace int with float to see the output
except ValueError as e:
    print("Error type:", type(e).__name__)
    print("Message:", e)
````

````python
try:
    value = float("12.5")
except ValueError as e:
    print("Error type:", type(e).__name__)
    print("Message:", e)
print(value)
````

#### Catching Any Exception with `Exception`

`except Exception as e:` catches almost all runtime exceptions while still letting you see what happened. It is safer than a bare `except:`.

````python
numbers = [10, 20, 30]

try:
    print(numbers[5])
except Exception as e:
    print("Something went wrong:", type(e).__name__, "-", e)
````
