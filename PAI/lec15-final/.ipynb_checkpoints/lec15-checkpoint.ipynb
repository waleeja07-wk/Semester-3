File Handling: File Modes, the with Statement, and Working with Paths

## 1. Definition: File Modes

File Modes specify how a file should be opened and what operations are allowed on that file.

#### Explanation

- Mode is the second argument of open().
- Determines read/write permissions.
- Different modes serve different purposes.
- Some modes create files automatically.
- Choosing the wrong mode may overwrite data.

#### General Syntax

```python
open("filename.txt", "mode")
```

## 2. Common File Modes

| Mode | Purpose |
|---|---|
| "r" | Read only |
| "w" | Write (overwrite) |
| "a" | Append |
| "x" | Create new file |
| "r+" | Read and write |
| "w+" | Write and read |
| "a+" | Append and read |

### Read Mode ("r")

#### Explanation

- Default mode.
- File must already exist.
- Cannot modify contents.
- Used for viewing data.

#### Example

````python
file = open("data.txt", "r")
print(file.read())
file.close()
````

### Write Mode ("w")

#### Explanation

- Creates file if missing.
- Deletes old contents.
- Writes new data.
- Useful for generating reports.

#### Example

````python
file = open("report.txt", "w")
file.write("Monthly Report")
file.close()
````

### Append Mode ("a")

#### Explanation

- Adds data at file end.
- Preserves existing contents.
- Useful for logs and records.
- Creates file if missing.

#### Example

````python
file = open("log.txt", "a")
file.write("User logged in\n")
file.close()
````

### Create Mode ("x")

#### Explanation

- Creates a new file.
- Generates error if file exists.
- Prevents accidental overwriting means create a new file, but only if that file does not already exist.
- Useful for unique files.

#### Example

````python
file = open("newfile.txt", "x")
file.close()
````

## 3. Definition: The with Statement

The `with` statement manages the file automatically. Even if the program encounters an error, Python closes the file without requiring you to write `file.close()` manually.

#### Explanation

- Eliminates manual close().
- Safer and cleaner code.
- Prevents resource leaks.
- Automatically closes files.
- Recommended Python practice.

### Traditional Method

````python
file = open("data.txt", "r")
print(file.read())
file.close()
````

### Using with

````python
with open("data.txt", "r") as file:
    print(file.read())
````

File closes automatically.

## 4. Advantages of the with Statement

#### Explanation

- Less code.
- Automatic cleanup.
- Better readability.
- Handles exceptions/errors safely.
- Recommended in professional Python development.

#### Example

````python
with open("students.txt", "w") as file:
    file.write("Ali")
````

No need for:

````python
file.close()
````

## 5. Definition: Working with Paths

A Path specifies the location of a file or folder in the file system.

#### Explanation

- Helps Python locate files.
- Can be absolute or relative.
- Different operating systems use different path styles.
- Python provides modules for handling paths.
- Proper path management improves portability.

## 6. Absolute Paths

#### Explanation

- Full location from root directory.
- Works regardless of current folder.
- Usually longer.
- Useful when exact location is known.

#### Example

Windows

``` python
file = open("C:/Users/Admin/Documents/data.txt", "r")
```


Linux

```python
file = open("/home/admin/data.txt", "r")
```

## 7. Relative Paths

#### Explanation

- Relative to current working directory.
- Shorter and more portable.
- Commonly used in projects.
- Easier to move projects.

#### Example

````python
file = open("data.txt", "r")
````

or

````python
file = open("files/data.txt", "r")
````

## 8. Using the `os` Module with Paths

#### Explanation

- Provides path-related utilities.
- Can determine current directory.
- Can create portable paths.
- Useful for file management.

#### Example

````python
import os

print(os.getcwd()) # Prints the Current Working Directory(CWD).
````

Output (example):

```
C:\Users\Admin
```

## 9. Using os.path.join()

#### Explanation

- Safely combines folder and file names.
- Works across operating systems.
- Avoids path separator issues. It joins file and directory names correctly, regardless of the operating system.
- Recommended for portability.

#### Example

````python
import os

path = os.path.join("data", "students.txt")
print(path)
````

Output (Windows):

```
data\students.txt
```

Output (Linux):

```
data/students.txt
```

```python
path = os.path.join("data", "CS", "Semester3", "students.txt")
print(path)
```

Output (Linux):

```
data/CS/Semester3/students.txt
```

## 10. Introduction to pathlib

#### Explanation

- Modern path-handling module.
- `pathlib` follows an object-oriented approach, allowing us to represent and manipulate file paths as objects using methods such as `exists()`, `read_text()`, and `write_text()`.
- Easier to read and maintain.
- Introduced in Python 3.

#### Example

````python
from pathlib import Path

# Creates a file path by joining the folder and file name.
file_path = Path("data") / "students.txt"

print(file_path)
````

## General Syntax

```python

Path("directory1") / "directory2" / "directory3" / "filename.txt"

```
